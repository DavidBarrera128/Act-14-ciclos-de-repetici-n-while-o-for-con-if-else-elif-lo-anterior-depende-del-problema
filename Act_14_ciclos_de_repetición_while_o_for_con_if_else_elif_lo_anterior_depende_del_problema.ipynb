### Act 14 ciclos de repetición while o for con if else elif lo anterior depende del problema
### Clase: Escribe un programa que pida el limite inferior y superior de un intervalo. Si el límite inferior es mayor que el superior lo tiene que volver a pedir. A continuación se van introduciendo números hasta que introduzcamos el 0. Cuando termine el programa dará las siguientes informaciones:

* La suma de los números que están dentro del intervalo (intervalo abierto).
* Cuantos números están fuera del intervalo.
* He informa si hemos introducido algún número igual a los límites del intervalo.


```python
# 1. Validación del intervalo
while True:
    limite_inferior = float(input("Introduce el límite inferior del intervalo: "))
    limite_superior = float(input("Introduce el límite superior del intervalo: "))

    if limite_inferior <= limite_superior:
        print("¡Intervalo correcto!\n")
        break
    else:
        print("Error: El límite inferior no puede ser mayor que el superior. Inténtalo de nuevo.\n")

# 2. Inicialización de variables de control
suma_dentro = 0.0
fuera_intervalo = 0
igual_a_limites = False

print("Introduce números (introduce el 0 para terminar):")

# 3. Bucle para introducir números
while True:
    numero = float(input("-> "))

    # Condición de salida
    if numero == 0:
        break

    # Clasificación del número
    if limite_inferior < numero < limite_superior:
        # Intervalo abierto (no incluye los límites)
        suma_dentro += numero
    elif numero == limite_inferior or numero == limite_superior:
        # El número es igual a uno de los límites
        igual_a_limites = True
    else:
        # El número está totalmente fuera del intervalo
        fuera_intervalo += 1

# 4. Mostrar los resultados finales
print("\n--- RESULTADOS ---")
print(f"La suma de los números dentro del intervalo abierto es: {suma_dentro}")
print(f"Cantidad de números fuera del intervalo: {fuera_intervalo}")

if igual_a_limites:
    print("Aviso: Se introdujo al menos un número igual a los límites del intervalo.")
else:
    print("No se introdujo ningún número igual a los límites del intervalo.")
print("Programa realizadopor David Barrera 1388")
```
